```python
choice=input('enter Your Choice(1-6)')
if choice == '1':
    print('Add Student Selected')
elif choice=='2':
    print('View Student Selected')
elif choice =='3':
    print('Search Student Selected')
elif choice =='4':
    print('Update Student Selected')
elif choice=='5':
    print('Delete Student Selected')
elif choice=='6':
    print('Exiting Programing')
else:
    print('Invalid Choice')
```

```python
students=[]
student_id=input('Enter student ID')
name=input('Enter Student Name')
age=input('Enter Student Age')
course=input('Enter Student Course')

student={
    'id': student_id,
    'name': name,
    'age': age ,
    'course': course
}
students.append(student)

print('Student Added Successfully')
print(student)
    
```

```python
if len(student)==0:
    print('No Student Selected ')
else:
    for student in students:
        print('___________________')
        print('Student_id',student['id'])
        print('name',student['name'])
        print('age',student['age'])
        print('course',student['course'])
```

```python
student_id=input('Enter Student ID to Search')

found = False
for student in students:
    if student['id']==student_id:
        print('Student Founded!')
        print('ID',student['id'])
        print('name',student['name'])
        print('age',student['age'])
        print('course',student['course'])
        found = True
        break
if found==False:
    print('Student Not Found')
```

```python
update_id=input('Enter Student ID to Update')

found = False

for student in students:
    if student['id']==update_id:
        
        print('Student Founded')

        student['name']=input('Enter student New name')
        student['age']=int(input('Enter Student new Age '))
        student['course']=input('Enter new Course')

        print('Student Updated Successfully')
        found = True
        break
        
if found == False:
      print('Student not found')
```

```python
delete_id = input('Enter Student Id to Dalete')

found = False 

for student in students:
    if student['id']==delete_id:
        students.remove(student)

        print('Student Deleted Succesfully')
        found = True
        break

if found == False:
    print('Student Not Found')
```

```python
import json

with open("students.json", "w") as file:
    json.dump(students, file, indent=4)
```

```python
import json

with open("students.json", "r") as file:
    students = json.load(file)

print("Student records loaded successfully!")
print(students)
```
